Модифицировать источник так, чтобы для определенной энергии испускалось случайное число фотонов. Число должно подчиняться распределению Пуассона со средним E*Y.
Реализовать метод генерации фотонов из задачи 2 как generator (конструкция в python).
Модифицировать генератор так, чтобы он возвращал (yield) по 100 фотонов вместо одного.

```python
import numpy as np

class Source:
    def __init__(self, Y: int):
        self.Y = Y

    def photon_directions(self, E: float):
       
        N = np.random.poisson(self.Y * E)  # число фотонов теперь испускается по распределению Пуассонасо средним значением E*Y

        phi = np.random.uniform(0, 2*np.pi, N)
        cosTheta = np.random.uniform(-1, 1, N)
        sinTheta = np.sqrt(1 - cosTheta**2)

        x = sinTheta * np.cos(phi)
        y = sinTheta * np.sin(phi)
        z = cosTheta

        return np.array([x, y, z]).T


source = Source(Y=10000)
```

Реализуем метод генерации фотонов 

```python
class Source:
    def __init__(self, Y: int):
        self.Y = Y

    def photon_directions(self, E: float):
        
        N = np.random.poisson(self.Y * E)

        # Генерируем направления и выдаём их по одному
        for i in range(N):
            phi = np.random.uniform(0, 2*np.pi)
            cosTheta = np.random.uniform(-1, 1)
            sinTheta = np.sqrt(1 - cosTheta**2)

            x = sinTheta * np.cos(phi)
            y = sinTheta * np.sin(phi)
            z = cosTheta

            yield np.array([x, y, z]) #


source = Source(Y=10000)
```

Модифицируем генератор так, чтобы он возвращал (yield) по 100 фотонов вместо одного

```python
class Source:
    def __init__(self, Y: int):
        self.Y = Y

    def photon_directions(self, E: float):
        N = np.random.poisson(self.Y * E)

        for i in range(0, N, 100): # генерируем направления фотонов по 100 а не по одному в отличие от предыдущей ячейки
            n = min(100, N - i)

            phi = np.random.uniform(0, 2*np.pi, n)
            cosTheta = np.random.uniform(-1, 1, n)
            sinTheta = np.sqrt(1 - cosTheta**2)

            x = sinTheta * np.cos(phi)
            y = sinTheta * np.sin(phi)
            z = cosTheta

            yield np.array([x, y, z]).T


source = Source(Y=10000)
```
